```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#  criando data frame para exemplo
dados = {'produtos': ['abacate', 'banana', 'maça', 'laranja', 'abacaxi', 'uva', 'melão'],
         'quantidade': [3, 4, 5, 6, 7, 7, 10]}

df = pd.DataFrame(dados)
```

```python
df.head(7)
```

```python
# verificando MEDIA dos dados

df['quantidade'].mean()
```

```python
# verificando MEDIANA dos dados

df['quantidade'].median()
```

```python
# verificando MODA dos dados

df['quantidade'].mode()
```

```python
# verificando VARIANCIA dos dados

# parametro ddof = 0 VARIANCIA POPULACIONAL
# parametro ddof = 1 VARIANCIA AMOSTRAL

df['quantidade'].var(ddof=0)
```

```python
# verificando DESVIO PADRÂO dos dados

df['quantidade'].std(ddof=0)
```

```python
# verificar QUARTIS no DataFrame

print('Primeiro quartil: ', df['quantidade'].quantile(0.25))
print('Segundo quartil: ', df['quantidade'].quantile(0.50))
print('Terceiro quartil: ', df['quantidade'].quantile(0.75))
print('Aplitude: ', df['quantidade'].max() - df['quantidade'].min())
```

```python
ax = sns.histplot(df.quantidade, stat="density", kde=True, color="blue")
```
